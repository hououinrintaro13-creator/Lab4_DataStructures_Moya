```python
# Step 2: Generate Performance Records, Categories, and Edge Cases
import random

# Student Identity & Parameters
LAST_NAME = "Moya"
STUDENT_ID = "TUPM-26-1138"
FAVORITE_ARTIST = "The Smiths"

# Derived computational parameters
seed_num = int(STUDENT_ID[-1])
id_checksum = sum(int(d) for d in STUDENT_ID if d.isdigit())

# Student & Subjects setup
student_names = [f"{LAST_NAME}_Student_{i}" for i in range(1, 6)]
subjects = [
    "Computer Programming",
    "Ethics",
    "Understanding the Self",
    "Chemistry for Engineers",
    "Differential Calculus"
]

# Seed the random number generator using your student seed
random.seed(seed_num)

performance_records = []

# Generate valid records for each student across subjects
for student in student_names:
    for subject in subjects:
        # Derive score using seed_num and character codes of FAVORITE_ARTIST
        char_sum = sum(ord(c) for c in FAVORITE_ARTIST)
        base_score = 75 + (char_sum % 20) + (seed_num % 5)
        score = min(100, max(70, base_score + random.randint(-8, 12)))
        
        # Determine performance category
        if score >= 90:
            category = "Excellent"
        elif score >= 85:
            category = "Above Average"
        elif score >= 78:
            category = "Satisfactory"
        else:
            category = "Passing"
            
        performance_records.append({
            "student": student,
            "subject": subject,
            "score": score,
            "category": category
        })

# 1. Include one intentional repeated student-subject record[cite: 22]
repeated_record = {
    "student": student_names[0],
    "subject": subjects[0],
    "score": performance_records[0]["score"],
    "category": performance_records[0]["category"]
}
performance_records.append(repeated_record)

# 2. Include one student record with a missing score (Incomplete)[cite: 22]
incomplete_record = {
    "student": student_names[-1],
    "subject": subjects[-1],
    "score": None,
    "category": "Incomplete"
}
performance_records.append(incomplete_record)

print("=== GENERATED PERFORMANCE RECORDS ===")
print(f"Total Records Generated (incl. edge cases): {len(performance_records)}")
print("Sample Record:", performance_records[0])
```

```python
# Step 3: Analyze Data, Handle Duplicates, and Print Consolidated Report
from collections import defaultdict

# 1. De-duplication using Sets (convert dictionaries to a tuple of items for set comparison)
unique_record_tuples = set(tuple(sorted(r.items())) for r in performance_records)
unique_records = [dict(t) for t in unique_record_tuples]

# Separate valid records from incomplete/repeated ones for analysis
valid_records = [r for r in unique_records if r["score"] is not None]
incomplete_records = [r for r in performance_records if r["score"] is None]

# 2. Summarize performance by student and by subject
student_summaries = defaultdict(list)
subject_summaries = defaultdict(list)

for r in valid_records:
    student_summaries[r["student"]].append(r["score"])
    subject_summaries[r["subject"]].append(r["score"])

print("==================================================")
print("       FINAL STUDENT PERFORMANCE REPORT           ")
print("==================================================\n")

print(f"Total Raw Records Processed: {len(performance_records)}")
print(f"Unique Valid Records: {len(valid_records)}")
print(f"Incomplete Records Flagged: {len(incomplete_records)}")
print("-" * 50)

print("\n--- STUDENT PERFORMANCE SUMMARIES ---")
for student, scores in student_summaries.items():
    avg_score = sum(scores) / len(scores)
    print(f"{student}: Average Score = {avg_score:.2f}")

print("\n--- SUBJECT PERFORMANCE SUMMARIES ---")
for subject, scores in subject_summaries.items():
    avg_score = sum(scores) / len(scores)
    print(f"{subject}: Average Score = {avg_score:.2f}")

print("\n--- FLAGGED INCOMPLETE RECORDS ---")
for inc in incomplete_records:
    print(f"Student: {inc['student']} | Subject: {inc['subject']} | Status: {inc['category']}")

print("==================================================")
```
