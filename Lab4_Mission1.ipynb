```python
# Mission 1: Engineering Test Data Organizer - Setup & Generation
import random

# Student Identity & Parameters
LAST_NAME = "Moya"
STUDENT_ID = "TUPM-26-1138"
FAVORITE_ARTIST = "The Smiths"

# Derived computational parameters
seed_num = int(STUDENT_ID[-1])

# 1. Generate five test sources using LAST_NAME with labels T1 to T5[cite: 11]
test_sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]

# 2. Define the five common measurement categories[cite: 11]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

# Seed the random generator
random.seed(seed_num)

engineering_records = []

# Generate measurement records across sources and categories[cite: 11]
for source in test_sources:
    for category in categories:
        # Generate numerical values using seed_num and character codes of FAVORITE_ARTIST[cite: 11]
        char_sum = sum(ord(c) for c in FAVORITE_ARTIST)
        base_val = (char_sum % 50) + (seed_num * 5)
        value = round(base_val + random.uniform(-10.0, 25.0), 2)
        
        engineering_records.append({
            "source": source,
            "category": category,
            "value": value
        })

# 3. Include one repeated test-source/category record[cite: 11]
repeated_record = {
    "source": test_sources[0],
    "category": categories[0],
    "value": engineering_records[0]["value"]
}
engineering_records.append(repeated_record)

# 4. Include multiple measurements under at least two categories[cite: 11]
extra_record_1 = {
    "source": test_sources[1],
    "category": categories[1],
    "value": 145.50
}
extra_record_2 = {
    "source": test_sources[2],
    "category": categories[2],
    "value": 48.20
}
engineering_records.extend([extra_record_1, extra_record_2])

print("=== GENERATED ENGINEERING TEST DATASET ===")
print(f"Total Records Generated (incl. edge cases): {len(engineering_records)}")
print("Sample Record:", engineering_records[0])
```

```python
# Step 2: Engineering Test Data Analysis, De-duplication, and Consolidated Summary
from collections import defaultdict

# 1. De-duplicate records using Sets
unique_eng_tuples = set(tuple(sorted(rec.items())) for rec in engineering_records)
unique_engineering_records = [dict(t) for t in unique_eng_tuples]

# 2. Summarize measurements by source and by category
source_summaries = defaultdict(list)
category_summaries = defaultdict(list)

for rec in unique_engineering_records:
    source_summaries[rec["source"]].append(rec["value"])
    category_summaries[rec["category"]].append(rec["value"])

# 3. Identify distinct entities
distinct_sources = set(rec["source"] for rec in unique_engineering_records)
distinct_categories = set(rec["category"] for rec in unique_engineering_records)

print("==================================================")
print("     FINAL ENGINEERING TEST DATA REPORT           ")
print("==================================================\n")

print(f"Total Raw Engineering Records Collected: {len(engineering_records)}")
print(f"Unique Distinct Engineering Records: {len(unique_engineering_records)}")
print(f"Duplicate/Repeated Records Flagged: {len(engineering_records) - len(unique_engineering_records)}")
print("-" * 50)

print("\n--- DISTINCT ENTITIES ---")
print(f"Test Sources ({len(distinct_sources)}): {sorted(list(distinct_sources))}")
print(f"Measurement Categories ({len(distinct_categories)}): {sorted(list(distinct_categories))}")

print("\n--- TEST SOURCE SUMMARIES ---")
for source, values in sorted(source_summaries.items()):
    avg_val = sum(values) / len(values)
    max_val = max(values)
    min_val = min(values)
    print(f"{source}: Readings = {len(values)} | Avg = {avg_val:.2f} | Min = {min_val:.2f} | Max = {max_val:.2f}")

print("\n--- MEASUREMENT CATEGORY SUMMARIES ---")
for category, values in sorted(category_summaries.items()):
    avg_val = sum(values) / len(values)
    max_val = max(values)
    min_val = min(values)
    print(f"{category}: Readings = {len(values)} | Avg = {avg_val:.2f} | Min = {min_val:.2f} | Max = {max_val:.2f}")

print("==================================================")
```
