```python
# Additional Exercise 2: Transaction History Analyzer - Setup & Generation
import random

# Student Identity & Parameters
LAST_NAME = "Moya"
STUDENT_ID = "TUPM-26-1138"
FAVORITE_ARTIST = "The Smiths"

# Derived computational parameters
seed_num = int(STUDENT_ID[-1])

# 1. Generate five customer names using LAST_NAME with labels C1 to C5[cite: 19]
customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]

# 2. Define the five common products[cite: 19]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]

# 3. Assign fixed prices to the five products[cite: 19]
product_prices = {
    "NOTEBOOK": 50.00,
    "PEN": 25.00,
    "USB": 350.00,
    "HEADPHONES": 1200.00,
    "CALCULATOR": 750.00
}

# Seed the random generator
random.seed(seed_num)

transaction_records = []

# Generate multiple transactions involving different customers and products[cite: 19]
for _ in range(12):
    customer = random.choice(customers)
    product = random.choice(products)
    
    # Generate quantity using seed_num and character codes of FAVORITE_ARTIST
    char_sum = sum(ord(c) for c in FAVORITE_ARTIST)
    quantity = max(1, (char_sum % 5) + (seed_num % 3) + random.randint(0, 2))
    
    transaction_records.append({
        "customer": customer,
        "product": product,
        "quantity": quantity,
        "price": product_prices[product],
        "total_value": quantity * product_prices[product]
    })

# Include at least one repeated customer-product transaction[cite: 19]
repeated_transaction = {
    "customer": customers[0],
    "product": products[0],
    "quantity": 2,
    "price": product_prices[products[0]],
    "total_value": 2 * product_prices[products[0]]
}
transaction_records.append(repeated_transaction)

print("=== GENERATED TRANSACTION DATASET ===")
print(f"Total Transactions Generated (incl. repeat): {len(transaction_records)}")
print("Sample Transaction:", transaction_records[0])
```

```python
# Step 2: Transaction Analysis, De-duplication, and Consolidated Report
from collections import defaultdict

# 1. De-duplicate transactions using Sets
unique_tx_tuples = set(tuple(sorted(tx.items())) for tx in transaction_records)
unique_transactions = [dict(t) for t in unique_tx_tuples]

# 2. Compute overall dataset transaction value
overall_total_value = sum(tx["total_value"] for tx in unique_transactions)

# 3. Summarize by customer and by product
customer_summaries = defaultdict(float)
product_summaries = defaultdict(int)
product_revenue = defaultdict(float)

for tx in unique_transactions:
    customer_summaries[tx["customer"]] += tx["total_value"]
    product_summaries[tx["product"]] += tx["quantity"]
    product_revenue[tx["product"]] += tx["total_value"]

# 4. Identify distinct customers and products
distinct_customers = set(tx["customer"] for tx in unique_transactions)
distinct_products = set(tx["product"] for tx in unique_transactions)

print("==================================================")
print("     FINAL CONSOLIDATED TRANSACTION REPORT        ")
print("==================================================\n")

print(f"Total Raw Transactions Collected: {len(transaction_records)}")
print(f"Unique Distinct Transactions: {len(unique_transactions)}")
print(f"Overall Dataset Transaction Value: PHP {overall_total_value:,.2f}")
print("-" * 50)

print("\n--- DISTINCT ENTITIES ---")
print(f"Distinct Customers ({len(distinct_customers)}): {sorted(list(distinct_customers))}")
print(f"Distinct Products ({len(distinct_products)}): {sorted(list(distinct_products))}")

print("\n--- CUSTOMER SPENDING SUMMARIES ---")
for customer, total_spent in customer_summaries.items():
    print(f"{customer}: Total Spent = PHP {total_spent:,.2f}")

print("\n--- PRODUCT SALES SUMMARIES ---")
for product in distinct_products:
    units = product_summaries[product]
    rev = product_revenue[product]
    print(f"{product}: Units Sold = {units} | Total Revenue = PHP {rev:,.2f}")

print("==================================================")
```
